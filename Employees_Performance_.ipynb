```python
import pandas as pd
data=pd.read_csv('/content/Uncleaned_employees_final_dataset (1).csv')
data
```

Sale & Marketing and Operations departments have the highest count

```python
import matplotlib.pyplot as plt
data['department'].value_counts().plot(kind='pie',autopct='%1.2f%%',shadow=True)
plt.show()
```

We have a strong relationship between the age and length of service

```python
import seaborn as sns
sns.pairplot(data)
```

The highest number of employees tenure is 5 years , right skewed

```python
import seaborn as sns
sns.histplot(data['length_of_service'])
```

There is a negative relationship between KPIs met more than 80 and lenght of service

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.regplot(x=data['length_of_service'],y=data['KPIs_met_more_than_80'],data=data,robust=True)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.hist(data['no_of_trainings'],bins=15)
plt.xlabel('no_of_trainings')
plt.ylabel('avg_training_score')
plt.show()
```
